```python
import numpy as np


betas = [0.9, 0.95]
final = []

rng = np.random.default_rng(42)
rwd = rng.uniform(size=(10,))
mean_rwd = np.mean(rwd)

t = 2
mean_state = [0.45, 0.23]
mean_state[0] = mean_state[0] * betas[0] + (1 - betas[0]) * mean_rwd
mean_state[1] = mean_state[1] * betas[1] + (1 - betas[1]) * mean_rwd**2

final.append((mean_state[0] / (1 - betas[0]**t), mean_state[1] / (1 - betas[1]**t)))

state = [0.45, 0.23]

for r in rwd:
    state[0] = state[0] * betas[0] + (1 - betas[0]) * r
    state[1] = state[1] * betas[1] + (1 - betas[1]) * r**2

final.append((state[0] / (1 - betas[0]**t), state[1] / (1 - betas[1]**t)))
final
```
