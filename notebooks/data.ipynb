```python
import random

import datasets


rng = random.seed(42)
data = datasets.load_dataset("parquet", data_files=["../data/train.parquet"])["train"]

sel_idxs = random.sample(list(range(len(data))), 10_000)
new_data = []

for idx in sel_idxs:
    new_data.append(data[idx])
```

```python
from datasets import Dataset, DatasetDict, Split


def gen(ds):
    for el in ds:
        yield el

train_ds = Dataset.from_generator(lambda: gen(new_data), split=Split.TRAIN)
final_ds = DatasetDict({"train": train_ds})
final_ds.push_to_hub("gh0stwin/deepscaler-subset")
```
